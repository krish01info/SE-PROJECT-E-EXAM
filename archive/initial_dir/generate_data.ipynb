```python
import os, random, itertools, datetime as dt
import pandas as pd
 
random.seed(42)
OUT = "./dataset"
SEC = os.path.join(OUT, "sections")
os.makedirs(SEC, exist_ok=True)
```

```python
COHORTS = [ # course id, program, year, sections, students per section
    ("BT2", "BTECH", 2, 9, 50), 
    ("BT3", "BTECH", 3, 9, 50),
    ("BB2", "BBA",   2, 3, 50),
    ("BB3", "BBA",   3, 3, 50),
]

secs = ["I" , "II" , "III" , "IV" , "V" , "VI" , "VII" , "VIII" , "IX"]
 
students = []          # student_id, cohort_id, section_id, roll_no
sections = []          # section_id, cohort_id, program, year, label, strength
for cid, prog, yr, nsec, size in COHORTS:
    seq = 1
    for s in range(nsec):
        sec_num = secs[s]
        sec_id = f"{cid}_{sec_num}"
        for k in range(size):
            sid = f"{cid}S{seq:04d}"
            students.append(dict(student_id=sid, cohort_id=cid,section_id=sec_id, roll_no=seq))
            seq += 1
        sections.append(dict(section_id=sec_id, cohort_id=cid, program=prog,year=yr, section_number=sec_num, strength=size))
 
students_df = pd.DataFrame(students)
sections_df = pd.DataFrame(sections)
students_df
```

```python
CORES = {
 "BT2": ["DSA","OOPS","LA","DAP","OS","DLD"],
 "BT3": ["ML","AI","SE","COA","IOT","CS_SKILL"],
 "BB2": ["MACROECON","BUSSTAT","BUSCOMM"],
 "BB3": ["BUSLAW","STRATMGMT","OPSMGMT","HAPPINESS"],
}
ELECTIVES = {
 "BT2": [("SEMIPHY",110), ("QM",90), ("DISCMATH",250)],
 "BT3": [("BIGDATA",115), ("CYBERSEC",95), ("ADVDS",240)],
 "BB2": [("FINANCE",55), ("MARKETING",55), ("DESTHINK",40)],
 "BB3": [("BIM",50), ("ETHICS",50), ("ORGBEH",50)],
}
TITLES = {
 "DSA":"Data Structures and Algorithms","OOPS":"Object Oriented Programming",
 "LA":"Linear Algebra","DAP":"Data Analysis in Python","OS":"Operating Systems",
 "DLD":"Digital Logic Design","ML":"Machine Learning","AI":"Artificial Intelligence",
 "SE":"Software Engineering","COA":"Computer Architecture and Organization",
 "IOT":"Internet of Things","CS_SKILL":"Coding Skills",
 "MACROECON":"Macroeconomics","BUSSTAT":"Business Statistics",
 "BUSCOMM":"Business Communication","BUSLAW":"Business Law",
 "STRATMGMT":"Strategic Management","OPSMGMT":"Operations Management",
 "HAPPINESS":"Happiness","SEMIPHY":"Semiconductor Physics","QM":"Quantum Mechanics",
 "DISCMATH":"Discrete Mathematics","BIGDATA":"Big Data Analysis",
 "CYBERSEC":"Cyber Security","ADVDS":"Advanced Data Science","FINANCE":"Finance",
 "MARKETING":"Marketing","DESTHINK":"Design Thinking",
 "BIM":"Bringing Ideas to Market","ETHICS":"Ethics and Integrity in Business",
 "ORGBEH":"Organizational Behaviour",
}
DEPT = {
 "DSA":"CSE","OOPS":"CSE","DAP":"CSE","OS":"CSE","ML":"CSE","AI":"CSE",
 "SE":"CSE","COA":"ECE","IOT":"ECE","DLD":"ECE","CS_SKILL":"CSE",
 "LA":"MATH","DISCMATH":"MATH","QM":"PHY","SEMIPHY":"PHY",
 "BIGDATA":"CSE","CYBERSEC":"CSE","ADVDS":"CSE",
 "MACROECON":"MGMT","BUSSTAT":"MGMT","BUSCOMM":"MGMT","BUSLAW":"LAW",
 "STRATMGMT":"MGMT","OPSMGMT":"MGMT","HAPPINESS":"HUM","FINANCE":"MGMT",
 "MARKETING":"MGMT","DESTHINK":"HUM","BIM":"MGMT","ETHICS":"LAW","ORGBEH":"HUM",
}
```

```python
courses = []   # course_id == paper_id here
for cid, _, _, _, _ in COHORTS:
    for c in CORES[cid]:
        courses.append(dict(course_id=f"{cid}_{c}", paper_id=f"{cid}_{c}",code=f"{cid}-{c}", title=TITLES[c], cohort_id=cid,kind="CORE", dept=DEPT[c]))
    for c, _ in ELECTIVES[cid]:
        courses.append(dict(course_id=f"{cid}_{c}", paper_id=f"{cid}_{c}",code=f"{cid}-{c}", title=TITLES[c], cohort_id=cid,kind="ELECTIVE", dept=DEPT[c]))
courses_df = pd.DataFrame(courses)
courses_df
```

```python
# Everyone takes every core of their cohort + exactly one elective,
# so the elective is the only thing worth storing per student.
# Shuffle first so electives cut across sections instead of following roll order.
rng = random.Random(42)   # local seed: cell is reproducible on its own

elective_of = {}
for cid, _, _, _, _ in COHORTS:
    pool = students_df[students_df.cohort_id == cid].student_id.tolist()
    rng.shuffle(pool)
    i = 0
    for c, n in ELECTIVES[cid]:
        for s in pool[i:i+n]:
            elective_of[s] = f"{cid}_{c}"
        i += n
    assert i == len(pool), (cid, i, len(pool))

students_df["elective_course_id"] = students_df.student_id.map(elective_of)
assert students_df.elective_course_id.notna().all()

def papers_of(cohort_id, elective_course_id):   # papers a student sits for
    return [f"{cohort_id}_{c}" for c in CORES[cohort_id]] + [elective_course_id]

students_df
```
